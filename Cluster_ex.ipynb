```python
import pandas as pd
df=pd.read_csv("/content/drive/MyDrive/FSP 3RD DEM/Mall_Customers.csv")
df
```

```python
x=df.iloc[:,[3,4]].values
x
```

```python
import scipy.cluster.hierarchy as sch
import matplotlib.pyplot as plt
dendrogram=sch.dendrogram(sch.linkage(x,method="ward"))
plt.xlabel("Customer")
plt.ylabel("Euclidean Distance")
plt.title("Dendrogram")
plt.show()
```

```python
from sklearn.cluster import KMeans
model=KMeans(n_clusters=5,random_state=0)
y_pred=model.fit_predict(x)
y_pred
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[y_pred==0,0],x[y_pred==0,1],s=100,c='coral',label='Cluster 1')
plt.scatter(x[y_pred==1,0],x[y_pred==1,1],s=100,c='purple',label='Cluster 2')
plt.scatter(x[y_pred==2,0],x[y_pred==2,1],s=100,c='grey',label='Cluster 3')
plt.scatter(x[y_pred==3,0],x[y_pred==3,1],s=100,c='cyan',label='Cluster 4')
plt.scatter(x[y_pred==4,0],x[y_pred==4,1],s=100,c='pink',label='Cluster 5')
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")
plt.title("Clusters of Customers")
plt.legend()
```
