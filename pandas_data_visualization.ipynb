```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5,6,7,8,9]
y=[1,4,9,16,25,36,49,64,81]
plt.figure(figsize=(10,2))
plt.plot(x,y,"-^",color='orange')
plt.grid()
plt.xlabel("x-values")
plt.ylabel("y-values")
plt.title("x-y dependency")
plt.show()
```
